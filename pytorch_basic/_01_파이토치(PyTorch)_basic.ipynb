**랜덤하게 가중치를 적용해 사인곡선 그리기**

```python
import math
import torch
import matplotlib.pyplot as plt

# -pi부터 pi 사이에서 점을 1,000개 추출
x = torch.linspace(-math.pi, math.pi, 1000)

# 실제 사인곡선에서 추출한 값으로 y 만들기
y = torch.sin(x)

# 예측 사인곡선에 사용할 임의의 가중치(계수)를 뽑아 y 만들기
a = torch.randn(())
b = torch.randn(())
c = torch.randn(())
d = torch.randn(())

# 사인 함수를 근사할 3차 다항식 정의
y_random = a * x**3 + b * x**2 + c * x + d

# 실제 사인곡선을 실제 y값으로 만들기
plt.subplot(2, 1, 1)
plt.title('y true')
plt.plot(x, y)

# 예측 사인곡선을 임의의 가중치로 만든 y값으로 만들기
plt.subplot(2, 1, 2)
plt.title('y pred')
plt.plot(x, y_random)

plt.show()
```

```python
import torch
import matplotlib.pyplot as plt


x = torch.tensor([150, 160, 170, 175, 180])
y = torch.tensor([50, 70, 64, 80, 75.])
N = len(x)
plt.scatter(x, y)
```

```python
# 초기값 설정
a = 0.45
b = -35
x_plot = torch.linspace(145, 190, 100)
y_plot = a * x_plot + b

plt.scatter(x, y)
plt.plot(x_plot, y_plot, color='red')
```

```python
# a, b를 바꿔가면서 Loss 값을 일일히 구해서 가장 작아지게 하는 a, b를 찾는 방법

a = 0.5 + torch.linspace(-0.2, 0.2, 100) # 0.3 ~ 0.7 사이의 100개 값
b = -30 + torch.linspace(-20, 20, 100) # -50 ~ -10 사이의 100개 값

L = torch.zeros(len(a), len(b)) # a, b의 조합에 따른 Loss 값을 저장할 행렬
for i in range(len(b)):
    for j in range(len(a)):
        for n in range(N):
            L[i, j] = L[i, j] + (y[n] - (a[j] * x[n] + b[i]))**2 # 기울기랑 절편을 바꿔가면서 Loss 값 계산
L = L / N # MSE
            
A, B = torch.meshgrid(a, b) # a, b의 좌표 행렬 생성
```

```python
print(torch.min(L), torch.argmin(L)) # 최소 Loss 값과 그 위치 출력

a_opt = A[L==torch.min(L)]
b_opt = B[L==torch.min(L)]

print(a_opt, b_opt) # 최소 Loss 값을 만드는 a, b 출력
```

```python
x_plot = torch.linspace(145, 190, 100)
y_plot = a_opt * x_plot + b_opt
plt.scatter(x, y)
plt.plot(x_plot, y_plot, color='red')
```
